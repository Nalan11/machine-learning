#### OneHotEncoding

This can be achieved by using any of the following 2 libraries:
1. Pandas - pd.get_dummies()
2. sklearn - OneHotEncoder()

Using Pandas

```python
weather_data_copy02 = weather_dataset.copy()
weather_data_copy02
```

```python
weather_data_copy02 = pd.get_dummies(data=weather_data_copy02, columns=['Weather'], dtype='int')
weather_data_copy02
```

Using sklearn

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
ohe = OneHotEncoder(sparse_output=False)
```

```python
weather_data_copy03 = weather_dataset.copy()
```

```python
weather_data_copy03.shape
```

```python
# weather_data_copy02['Weather'].array.reshape(-1, 1)

encoded = ohe.fit_transform(weather_data_copy03[['Weather']])

print(encoded[:10], '...\n\n')
print('shape:', encoded.shape)
```

```python
ohe.get_feature_names_out()
```

```python
encoded_df = pd.DataFrame(data=encoded, columns=ohe.get_feature_names_out())
encoded_df
```

```python
weather_data_copy03
```

```python
# Reset index and drop the old one
weather_data_copy03 = weather_data_copy03.reset_index(drop=True)

# The incorrect indexes make the concat misalign !!!
# Thus, need to reset the index as above
```

```python
pd.concat([weather_data_copy03.drop(columns=['Weather']), encoded_df], axis=1)
# x['Weather_C'].isnull()
```
